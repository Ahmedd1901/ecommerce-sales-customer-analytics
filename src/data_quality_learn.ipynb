```python
import pandas as pd
file_path = "C:/Users/ahmed/Desktop/Ecommerce-Sales-Customer-Analytics/data/online_retail_II.xlsx"

df = pd.read_excel(file_path)
```

```python
## 1. cancelled invoices
cancelled = df["Invoice"].astype(str).str.startswith("C")

print("Cancelled invoices:")
print(cancelled.sum())
```

```python
## 2. Negative quantities
print("Negative quantities:")
print((df["Quantity"] < 0).sum())
```

```python
## 3. Zero quantities
print("Zero quantities:")
print((df['Quantity'] == 0).sum())
```

```python
## 4. Zero or negative prices
p = (df['Price'] == 0).sum()
p1 = (df['Price'] < 0).sum()
print(f"Zero prices: {p}")
print(f"negative prices: {p1}")
```

```python
## 5. Rows with negative quantity but normal invoice
print(((df["Quantity"] < 0) & (~cancelled)).sum())
```

```python
# 6. Duplicate rows
print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
# 7. Missing customer IDs
print("Missing Customer IDs:")
print(df["Customer ID"].isna().sum())
```

```python
# 8. Missing descriptions
print("Missing descriptions:")
print(df["Description"].isna().sum())
```

```python
print("Rows with negative prices:")
print(df[df["Price"] < 0][
    ["Invoice", "StockCode", "Description", "Quantity", "Price", "Customer ID"]
].head(20))
```

```python
print("Rows with zero prices:")
print(df[df["Price"] == 0][
    ["Invoice", "StockCode", "Description", "Quantity", "Price", "Customer ID"]
].head(20))
```
