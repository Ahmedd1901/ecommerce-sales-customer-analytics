```python
import pandas as pd
file_path = "C:/Users/ahmed/Desktop/Ecommerce-Sales-Customer-Analytics/data/online_retail_II.xlsx"

df = pd.read_excel(file_path)
```

```python
cancelled = df["Invoice"].astype(str).str.startswith("C")

print("Total rows:", len(df))

print("Cancelled rows:", cancelled.sum())
print("Cancelled %:", round(cancelled.mean() * 100, 2), "%")
```

```python
negative_quantity = df["Quantity"] < 0

print("Negative quantity rows:", negative_quantity.sum())
print(
    "Negative quantity %:",
    round(negative_quantity.mean() * 100, 2),
    "%"
)
```

```python
zero_quantity = df["Quantity"] == 0

print("Zero quantity rows:", zero_quantity.sum())
```

```python
zero_price = df["Price"] == 0

print("Zero price rows:", zero_price.sum())
print(
    "Zero price %:",
    round(zero_price.mean() * 100, 2),
    "%"
)
```

```python
negative_price = df["Price"] < 0

print("Negative price rows:", negative_price.sum())
print(
    "Negative price %:",
    round(negative_price.mean() * 100, 2),
    "%"
)
```

```python
missing_customer = df["Customer ID"].isna()

print("Missing customer IDs:", missing_customer.sum())
print(
    "Missing customer ID %:",
    round(missing_customer.mean() * 100, 2),
    "%"
)
```

```python
duplicates = df.duplicated()

print("Duplicate rows:", duplicates.sum())
print(
    "Duplicate %:",
    round(duplicates.mean() * 100, 2),
    "%"
)
```

```python
print("Negative quantities by invoice type:")

print(
    pd.crosstab(
        negative_quantity,
        cancelled,
        rownames=["Negative Quantity"],
        colnames=["Cancelled Invoice"]
    )
)
```

```python
print("Example duplicate rows:")

duplicates_df = df[df.duplicated(keep=False)]

print(
    duplicates_df
    .sort_values(
        ["Invoice", "StockCode", "Description"]
    )
    .head(20)
)
```

```python
print("Duplicate invoice + product combinations:")

invoice_product_duplicates = (
    df.groupby(["Invoice", "StockCode"])
      .size()
      .sort_values(ascending=False)
)

print(
    invoice_product_duplicates[
        invoice_product_duplicates > 1
    ].head(20)
)
```

```python
# Issue	                   Decision
# Negative prices	        Remove
# Cancelled invoices	    Remove from sales dataset
# Negative quantities	    Investigate
# Zero prices	            Keep for now
# Missing Customer ID	    Keep for sales, exclude from customer analysis
# Missing Description	    Keep
# Exact duplicates	        Investigate, then likely remove
# Zero quantity	            None
```
