```python
import pandas as pd
df = pd.read_csv('../data/cleaned_retail.csv')
df.head()
```

```python
## Sales-only data
sales = df[df['Quantity'] > 0].copy()
```

```python
## Sales metrics
total_revenue = sales['Revenue'].sum()
total_units = sales['Quantity'].sum()
total_orders = sales['Invoice'].nunique()
average_order_value = total_revenue / total_orders
total_products = sales['StockCode'].nunique()
total_customers = sales["Customer ID"].nunique()
```

```python
print("Total Revenue:", total_revenue)

print("Total Units Sold:", total_units)

print("Total Orders:", total_orders)

print("Average Order Value:", average_order_value)

print("Unique Products:", total_products)

print("Unique Customers:", total_customers)
```

```python
## save sales data
sales.to_csv("../data/sales.csv", index=False)
```
