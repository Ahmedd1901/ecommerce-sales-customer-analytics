```python
import pandas as pd


file_path = "C:/Users/ahmed/Desktop/Ecommerce-Sales-Customer-Analytics/data/online_retail_II.xlsx"

df = pd.read_excel(file_path)
print("Original shape:",df.shape)
```

```python
## 1- Removing exact duplicates
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)
```

```python
## 2- removing cancelled invoices
cancelled = df['Invoice'].astype(str).str.startswith('C')
df = df[~cancelled]
print("After removing cancelled invoices:",df.shape)
```

```python
## 3- Removing negative prices
df = df[df['Price'] >= 0]
print("After removing negative prices:",df.shape)
```

```python
## 4- Create revenue
df['Revenue'] = df['Quantity'] * df['Price']
```

```python
## 5- Display results
print("Cleaned dataset:")
print(df.head())

print("Final shape:")
print(df.shape)

print("Missing values:")
print(df.isnull().sum())
```

```python
## We didn't remove negative quantities because they
## may represent returns/adjustments and those can
## be analytically useful
```

```python
df.to_csv("../data/cleaned_retail.csv", index=False)
```
